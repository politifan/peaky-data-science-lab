# SQL: группировка в отдельной памяти
Используйте после объяснения соответствующего модуля, не в первом уроке.

```python
import sqlite3
c=sqlite3.connect(":memory:")
c.execute("CREATE TABLE t(city TEXT, amount REAL)")
c.executemany("INSERT INTO t VALUES (?, ?)", [("A",2),("A",3),("B",7)])
print(c.execute("SELECT city,SUM(amount) FROM t GROUP BY city ORDER BY city").fetchall())
c.close()
```

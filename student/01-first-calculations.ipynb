## Цикл с явным карантином
Предскажите вывод до запуска. Затем измените один свой вход и повторите.

```python
values = [0, None, 5, -2]
valid = []
quarantine = []
for value in values:
    if value is not None and value >= 0:
        valid.append(value)
    else:
        quarantine.append(value)
print(len(valid), sum(valid))
print(quarantine)
```

## Функция и различающая проверка
Предскажите вывод до запуска. Затем измените один свой вход и повторите.

```python
def total_known(values):
    total = 0
    for value in values:
        if value is not None and value >= 0:
            total += value
    return total

assert total_known([0, None, 5, -2]) == 5
assert total_known([]) == 0
assert total_known([0]) == 0
```

## Категории и счётчики
Предскажите вывод до запуска. Затем измените один свой вход и повторите.

```python
counts = {}
for name in ["A", "a", "A"]:
    counts[name] = counts.get(name, 0) + 1
assert counts == {"A": 2, "a": 1}
print(counts)
```

## Сохранность строк по id
Предскажите вывод до запуска. Затем измените один свой вход и повторите.

```python
rows = [{"id":"a", "amount":0}, {"id":"b", "amount":None}, {"id":"c", "amount":5}, {"id":"d", "amount":-2}]
valid = []
quarantine = []
for row in rows:
    value = row["amount"]
    if value is not None and value >= 0:
        valid.append(row)
    else:
        quarantine.append(row)
assert {r["id"] for r in valid} == {"a", "c"}
assert {r["id"] for r in valid}.isdisjoint({r["id"] for r in quarantine})
assert {r["id"] for r in valid + quarantine} == {r["id"] for r in rows}
print(sum(r["amount"] for r in valid))
```
